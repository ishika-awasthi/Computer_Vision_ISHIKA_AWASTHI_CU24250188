2. Write a program to convert a color image into Grayscale and Binary Image.

```python
import cv2
from google.colab.patches import cv2_imshow

# Convert to grayscale
grayscale_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Display the grayscale image
print('Grayscale Image:')
cv2_imshow(grayscale_img)

# Save the grayscale image
cv2.imwrite("grayscale_output.jpeg", grayscale_img)
```

```python
# Convert to binary image (using a simple threshold)
# The threshold value (127) and max_value (255) can be adjusted.
# cv2.THRESH_BINARY creates a binary image where pixels above threshold are set to max_value (255) and below to 0.
# The first return value (ret) is the threshold value used.
ret, binary_img = cv2.threshold(grayscale_img, 127, 255, cv2.THRESH_BINARY)

# Display the binary image
print('Binary Image:')
cv2_imshow(binary_img)

# Save the binary image
cv2.imwrite("binary_output.jpeg", binary_img)
```

3. Write a program to compute and display the complement(negative) of an image.

```python
import cv2
from google.colab.patches import cv2_imshow

# Read the image (ensure 'img' is defined)
img = cv2.imread("input.jpeg")

# Compute the negative (complement) of the original color image
# For an 8-bit image, cv2.bitwise_not() inverts every bit of each pixel, effectively creating a negative.
negative_img = cv2.bitwise_not(img);

# Display the negative image
print('Negative (Complement) Image:')
cv2_imshow(negative_img);

# Save the negative image
cv2.imwrite("negative_output.jpeg", negative_img)
```

4. resize , crop and rotate image

### 4.1 Resizing an Image

```python
import cv2
from google.colab.patches import cv2_imshow

# --- Resize Image ---
# Define new dimensions (width, height)
# For example, resize to half its original size or a fixed size.
# Let's resize it to a fixed width of 400 pixels while maintaining aspect ratio.

original_height, original_width = img.shape[:2]
new_width = 400
ratio = new_width / original_width
new_height = int(original_height * ratio)

resized_img = cv2.resize(img, (new_width, new_height), interpolation = cv2.INTER_AREA)

# Display the resized image
print('Resized Image:')
cv2_imshow(resized_img)

# Save the resized image
cv2.imwrite("resized_output.jpeg", resized_img)
```

### 4.2 Cropping an Image

```python
import cv2
from google.colab.patches import cv2_imshow

# --- Crop Image ---
# Define the region of interest (ROI) for cropping: [start_row:end_row, start_column:end_column]
# Let's crop the center of the image. For demonstration, we'll crop a 200x200 square from the middle.

height, width = img.shape[:2]

start_row, start_col = int(height * .25), int(width * .25)
end_row, end_col = int(height * .75), int(width * .75)

cropped_img = img[start_row:end_row, start_col:end_col]

# Display the cropped image
print('Cropped Image:')
cv2_imshow(cropped_img)

# Save the cropped image
cv2.imwrite("cropped_output.jpeg", cropped_img)
```

### 4.3 Rotating an Image

```python
import cv2
from google.colab.patches import cv2_imshow

# --- Rotate Image ---
# Get the image dimensions
(h, w) = img.shape[:2]

# Calculate the center of the image
center = (w // 2, h // 2)

# Define the rotation angle and scale factor
angle = 45 # Rotate by 45 degrees
scale = 1.0 # No scaling

# Get the rotation matrix
M = cv2.getRotationMatrix2D(center, angle, scale)

# Perform the rotation
rotated_img = cv2.warpAffine(img, M, (w, h))

# Display the rotated image
print('Rotated Image (45 degrees):')
cv2_imshow(rotated_img)

# Save the rotated image
cv2.imwrite("rotated_output.jpeg", rotated_img)
```

### 5. Split and Merge Image Channels

```python
import cv2
from google.colab.patches import cv2_imshow

# --- Split Image Channels ---
# The img variable should still hold the original color image.
# OpenCV stores images in BGR format, so the channels are Blue, Green, Red.
(b, g, r) = cv2.split(img)

# Display individual channels
# Note: cv2_imshow expects a 3-channel image for full color or 1-channel for grayscale.
# To display single channels, we'll show them as grayscale (intensity of that channel).
print('Blue Channel:')
cv2_imshow(b)

print('Green Channel:')
cv2_imshow(g)

print('Red Channel:')
cv2_imshow(r)
```

```python
import cv2
from google.colab.patches import cv2_imshow

# --- Merge Image Channels ---
# Merge the channels back together in the original BGR order
merged_img = cv2.merge([b, g, r])

# Display the merged image
print('Merged Image:')
cv2_imshow(merged_img)

# Save the merged image (should be identical to the original color image if not modified)
cv2.imwrite("merged_output.jpeg", merged_img)
```

### 6. Split Image by Size (Tiling)

```python
import cv2
from google.colab.patches import cv2_imshow

# Read the image (ensure 'img' is defined from previous steps)
img = cv2.imread("input.jpeg")

# Get dimensions of the image
height, width, _ = img.shape

# Define tile size (e.g., 100x100 pixels)
tile_height = 100
tile_width = 100

print(f"Original image dimensions: {width}x{height}")
print(f"Splitting into tiles of size: {tile_width}x{tile_height}")

tile_count = 0
# Loop through the image to create tiles
for y in range(0, height, tile_height):
    for x in range(0, width, tile_width):
        # Define the coordinates for the current tile
        y1 = y
        y2 = y + tile_height
        x1 = x
        x2 = x + tile_width

        # Ensure we don't go out of bounds for the last tiles
        if y2 > height: y2 = height
        if x2 > width: x2 = width

        # Crop the tile
        tile = img[y1:y2, x1:x2]

        # Display the tile
        if tile.shape[0] > 0 and tile.shape[1] > 0: # Ensure the tile is not empty
            tile_count += 1
            print(f"Displaying Tile {tile_count} (Region: {x1}:{x2}, {y1}:{y2})")
            cv2_imshow(tile)
            # Optionally save each tile
            # cv2.imwrite(f"tile_{tile_count}.jpeg", tile)

print(f"Successfully split and displayed {tile_count} tiles.")
```

### 7. Merge Image from Tiles

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Re-read the original image to ensure we have the base to simulate tile content
# In a real scenario, you would load individual tile files here.
original_img = cv2.imread("input.jpeg")

# Get dimensions of the original image
height, width, channels = original_img.shape

# Define the tile size (same as used for splitting)
tile_height = 100
tile_width = 100

# Create an empty canvas to reconstruct the image
merged_img_from_tiles = np.zeros((height, width, channels), dtype=np.uint8)

print("Merging tiles back into a single image...")

# Loop through the image to place tiles back
for y in range(0, height, tile_height):
    for x in range(0, width, tile_width):
        # Define the coordinates for the current tile
        y1 = y
        y2 = y + tile_height
        x1 = x
        x2 = x + tile_width

        # Ensure we don't go out of bounds for the last tiles
        if y2 > height: y2 = height
        if x2 > width: x2 = width

        # Extract the corresponding tile from the original image
        # (In a real application, you would load this from a saved tile file)
        current_tile = original_img[y1:y2, x1:x2]

        # Place the tile into the merged image canvas
        merged_img_from_tiles[y1:y2, x1:x2] = current_tile

# Display the merged image
print('Reconstructed Merged Image from Tiles:')
cv2_imshow(merged_img_from_tiles)

# Save the merged image
cv2.imwrite("merged_from_tiles_output.jpeg", merged_img_from_tiles)

print("Image successfully merged and saved as merged_from_tiles_output.jpeg")
```

### 8. Split Image into Four Quarters

```python
import cv2
from google.colab.patches import cv2_imshow

# Read the image (ensure 'img' is defined from previous steps)
img = cv2.imread("input.jpeg")

# Get dimensions of the image
height, width, _ = img.shape

# Define tile size for splitting into four (2x2 grid)
tile_height = height // 2
tile_width = width // 2

print(f"Original image dimensions: {width}x{height}")
print(f"Splitting into 4 tiles of size: {tile_width}x{tile_height}")

tile_count = 0
# Loop through the image to create tiles
for y in range(0, height, tile_height):
    for x in range(0, width, tile_width):
        # Define the coordinates for the current tile
        y1 = y
        y2 = y + tile_height
        x1 = x
        x2 = x + tile_width

        # Crop the tile
        tile = img[y1:y2, x1:x2]

        # Display the tile
        if tile.shape[0] > 0 and tile.shape[1] > 0:
            tile_count += 1
            print(f"Displaying Quarter Tile {tile_count} (Region: {x1}:{x2}, {y1}:{y2})")
            cv2_imshow(tile)
            # Optionally save each tile
            cv2.imwrite(f"quarter_tile_{tile_count}.jpeg", tile)

print(f"Successfully split and displayed {tile_count} quarter tiles.")
```

### 9. Merge Image from Four Quarters

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Load the four quarter tiles
tile1 = cv2.imread("quarter_tile_1.jpeg") # Top-Left
tile2 = cv2.imread("quarter_tile_2.jpeg") # Top-Right
tile3 = cv2.imread("quarter_tile_3.jpeg") # Bottom-Left
tile4 = cv2.imread("quarter_tile_4.jpeg") # Bottom-Right

# Check if tiles were loaded successfully
if tile1 is None or tile2 is None or tile3 is None or tile4 is None:
    print("Error: One or more quarter tiles could not be loaded. Please ensure they were saved correctly.")
else:
    # Get dimensions of a single tile
    tile_height, tile_width, channels = tile1.shape

    # Calculate the dimensions of the full merged image
    full_height = tile_height * 2
    full_width = tile_width * 2

    # Create an empty canvas to reconstruct the image
    merged_quarters_img = np.zeros((full_height, full_width, channels), dtype=np.uint8)

    # Place each tile into its correct position
    merged_quarters_img[0:tile_height, 0:tile_width] = tile1 # Top-Left
    merged_quarters_img[0:tile_height, tile_width:full_width] = tile2 # Top-Right
    merged_quarters_img[tile_height:full_height, 0:tile_width] = tile3 # Bottom-Left
    merged_quarters_img[tile_height:full_height, tile_width:full_width] = tile4 # Bottom-Right

    # Display the merged image
    print('Merged Image from Four Quarters:')
    cv2_imshow(merged_quarters_img)

    # Save the merged image
    cv2.imwrite("merged_quarters_output.jpeg", merged_quarters_img)

    print("Image successfully merged from four quarters and saved as merged_quarters_output.jpeg")
```
